## Initial EDA Observations — 2026-10-02

### Basic facts
- postings: 123,849 rows × 31 columns
- job_skills: 213,768 rows (job_id → skill_abr mapping)
- skills: 35 rows (skill_abr → skill category name)
- postings_with_skills (after merge): 207,531 rows × 33 columns

### Salary data
- Salary present in only 24.06% of postings (~76% MISSING)
- Salary outliers observed:
  - max_salary max = 120,000,000 (unrealistic)
  - min_salary min = 1 (unrealistic)
  - normalized_salary max = 535,600,000 (unrealistic)
- Cleanup required in Phase 3 (SRS Section 4.4)

### Locations & titles
- Unique companies: 24,428
- Top locations: United States (8125), New York NY (2756), Chicago IL (1834)
- Top job titles: Sales Manager (673), Customer Service Rep (373), Project Manager (354)

### ⚠️ IMPORTANT — skills in this dataset are CATEGORIES, not specific skills
- skills.csv contains 35 BROAD categories:
  Information Technology, Sales, Management, Manufacturing,
  Health Care Provider, Business Development, Engineering, Other,
  Finance, Marketing, Accounting/Auditing, Administrative, ...
- These are NOT specific skills like Python, SQL, JavaScript.
- For JMIS's specific-skill analysis (Skill Demand Score, resume gaps,
  job recommendations), we will extract specific skills from the
  job `description` column in Phase 5 using a custom skills dictionary.
- Both levels will be used:
  * Coarse categories → industry/department level trends
  * Fine skills (Phase 5) → user-facing recommendations

### Data source summary
- postings.csv: main job records
- jobs/job_skills.csv: job_id → skill_abr (category code)
- mappings/skills.csv: skill_abr → skill category name
- skills_desc column is 98% empty (2,439 / 123,849) — DO NOT use for skill extraction

```python
import pandas as pd

job_skills = pd.read_csv("../data/raw/archive/jobs/job_skills.csv")
skills = pd.read_csv("../data/raw/archive/mappings/skills.csv")

print("job_skills:", job_skills.shape)
print("skills:", skills.shape)
```

```python
job_skills.head()
```

```python
skills.head()
```

```python
job_skills.info()
skills.info()
```

```python

job_skills_named = job_skills.merge(skills, on="skill_abr", how="left")

print("job_skills_named:", job_skills_named.shape)
job_skills_named.head()
```

```python

postings_with_skills = postings.merge(job_skills_named, on="job_id", how="left")

print("postings_with_skills:", postings_with_skills.shape)
postings_with_skills.head(20)
```
